# Jupyter to PDF Conversion Demo 🚀
Welcome to the **RenderPage** VS Code extension demo. This extension compiles Jupyter Notebooks (`.ipynb`) directly into high-fidelity, styled PDF documents **100% locally and offline** using your system's Google Chrome or Microsoft Edge browser.

It follows **SOLID**, **DRY**, and **KISS** principles to provide a lightweight, efficient, and robust tool.

$$f(x \mid \mu, \sigma^2) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}$$

We also support systems of equations:

$$\begin{cases} x + y = 10 \\ 2x - y = 5 \end{cases}$$

And standard matrices:

$$\mathbf{A} = \begin{pmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{pmatrix}$$

## 3. Matplotlib and Seaborn Plots 📈
Jupyter stores plotting outputs as base64-encoded images. Our renderer embeds them seamlessly, auto-fitting them to the margins, and guaranteeing page-break isolation so a plot never gets sliced in half across pages!

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x) * np.exp(-x/3)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label='Damped Sine', color='#0284c7', linewidth=2.5)
plt.title('Damped Sine Wave (Simulated)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
