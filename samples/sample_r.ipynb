This notebook demonstrates standard cells rendered in R when the notebook primary language is set to R.

```r
# R code sample
calculate_means <- function(data_vector) {
  # Clean and calculate mean
  clean_data <- na.omit(data_vector)
  mean_val <- mean(clean_data)
  return(mean_val)
}

scores <- c(88, 92, NA, 79, 95)
print(calculate_means(scores))
```
